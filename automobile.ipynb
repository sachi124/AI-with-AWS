#### Let's make the column header small letter and a-a to a_a

```python
automobile.columns = (
    automobile.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "", regex=False)
    .str.replace("-", "_", regex=False)
)

automobile.columns.tolist()
```

#### Phase 7: Convert it into numeric columns

```python
numeric_columns = [
    "normalized_losses",
    "wheel_base",
    "length",
    "width",
    "height",
    "curb_weight",
    "engine_size",
    "bore",
    "stroke",
    "compression_ratio",
    "horsepower",
    "peak_rpm",
    "city_mpg",
    "highway_mpg",
    "price"
]

for col in numeric_columns:
    automobile[col] = pd.to_numeric(automobile[col], errors='coerce')

automobile[numeric_columns].dtypes
```

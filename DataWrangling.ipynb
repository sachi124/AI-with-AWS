```python
# create total stats
df["sachin_bist"] = (
    df["hp"]
    + df["attack"]
    + df["defense"]
    + df["speed"]
    + df["special_attack"]
    + df["special_defense"]
)

print(
    df[
        ["species",
        "hp",
        "attack",
        "defense",
        "speed",
        "special_attack",
        "special_defense",
        "sachin_bist"]
    ].head()
)
```

```python
top_10_total = (
    df.sort_values("sachin_bist", ascending=False).head(10)
)
print(top_10_total[["species", "sachin_bist"]])
```

```python
stat_column = [
    "hp",
    "attack",
    "speed",
    "defense",
    "special_attack",
    "special_defense"
]

df["sachin_bista"] = df[stat_column].sum(axis=1)
df.head()
```

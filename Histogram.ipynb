#### Histogram

```python
# Import the necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
# Read the CSV file into a DataFrame
df = pd.read_csv('pokemon.csv')
df.head()
df.shape
df.info()
df.describe(include='all')
```

```python
df.head(10)
```

#### Example 1: Plot the default Histogram

```python
plt.hist(data=df, x='speed')
plt.show()
```

```python
# Example 2: histogram with fixed number of bins , i.e 20 bins
plt.hist(data=df, x='speed', bins=20)
plt.show()
```

#### histogram with a dynamic number of bins, using arange - Func

```python
# Create a bins wiht step size of 5
bins = np.arange(0, df['speed'].max() + 5, 5)

plt.hist(data=df, x='speed', bins=bins)
plt.show()
```

#### Alternative Approach: histplot()

```python
# Plot a histogram with KDE (Kernel Density Estimate) line
sns.histplot(data=df, x='speed', kde=True, stat='density', bins=20)
plt.show()
```

```python
# Define bin edges for the histogram based on the 'speed' column
bin_edges = np.arange(0, df['speed'].max() + 1, 5)

# Plot the distribution of the 'speed' column using seaborn
sns.histplot(data=df, x='speed', bins=bin_edges, alpha=1)
plt.show()
```

#### Plot Two Histogram side by side

```python
# Resize the chart, and have two plots side-by-side
# Set a larger figure size for subplots

plt.figure(figsize=[20, 5])

# Histogram on the left: example of too-large bin size
# 1 row, 2 columns, subplot 1
plt.subplot(1, 2, 1)
bins = np.arange(0, df['speed'].max() +4, 4)
plt.hist(data=df, x='speed', bins=bins)

# Histogram on the right: example of too-small bin size
# 1 row, 2 columns, subplot 2
plt.subplot(1, 2, 2)
bins = np.arange(0, df['speed'].max() + 0.25, 0.25)
plt.hist(data=df, x='speed', bins=bins)
plt.show()
```

```python
# Plot the smple pie charts
sorted_counts = df['generation_id'].value_counts()

plt.pie(sorted_counts, labels=sorted_counts.index, startangle=90, counterclock=False);
plt.axis('equal')
plt.show()
```

#### Doughnut Plot

```python
sorted_counts = df['generation_id'].value_counts()
plt.pie(sorted_counts, labels=sorted_counts.index, counterclock=False, startangle=90, wedgeprops={'width': 0.5, 'edgecolor': 'w'})
plt.axis('equal')
plt.show()
```

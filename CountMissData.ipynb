## Counting Missing Data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

# Read the data from csv
sachin_df = pd.read_csv("sales-data.csv")
sachin_df.head(10)
```

```python
sachin_df.shape
```

```python
# Use either function below: isna() or isnull()
sachin_df.isnull()
```

```python
# lets count missing value of each column
sachin_df.isna().sum()
```

# prepare a NaN tabular data

```python
# Let's drop the column that do not have any NaN/None value

na_counts = sachin_df.drop(['Date', 'Temperature', 'Fuel_Price'], axis=1).isna().sum()
print(na_counts)
```

# Step 3 - Plot the bar chart from the NaN tabular data, and also print values on each bar

```python
# The first argument to the function below contains the x-values (column names), the second argument the y-values (our counts).
# Refer to the syntax and more example here - https://seaborn.pydata.org/generated/seaborn.barplot.html

sns.barplot(x=na_counts.index.values, y=na_counts)

# get the current tick location labels
plt.xticks(rotation=90)

# logic to print value in each bar
for i in range (na_counts.shape[0]):
    count = na_counts[i]

    plt.text(i, count+300, count, ha = 'center', va = 'top')
```

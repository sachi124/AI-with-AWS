### Using a bar chart, demonstrate if instructors teaching lower-division courses receive higher average teaching evaluations.

```python
sachin_df.head()
```

```python
# find the average teaching evaluation in both groups of upper and lower division
division_eval = sachin_df.groupby('division')[['eval']].mean().reset_index()
```

```python
# Plot the barplot using the seaborn library
sns.set(style="whitegrid")
sx = sns.barplot(x="division", y="eval", data=division_eval)
plt.show()
```

```python
# Create a scatterplot with the scatterplot function in the seaborn library
ax = sns.scatterplot(x="age", y="eval", data=sachin_df)
plt.show()
```

### Using gender-differentiated scatter plots, plot the relationship between age and teaching evaluation scores.


```python
# Create a scatterplot with the scatterplot function in the seaborn lib this time add the hue arg
ax = sns.scatterplot(x="age", y="eval", hue="gender", data=sachin_df)
plt.show()
```

### Create a group of Histogram of taught by gender and tenure

```python
# We add Hue= tenure arg
sns.catplot(x="gender", hue="tenure", kind="count", data=sachin_df)
plt.show()
```

```python
# We add another args named row and use the division varaible as the row
sns.catplot(x='gender', hue='tenure', row='division', kind='count', data=sachin_df)
plt.show()
```
